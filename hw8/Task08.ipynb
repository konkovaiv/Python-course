Получить информацию обо всех книгах и сохранить в pd.DataFrame




```python
import requests
import pandas as pd
```

```python
api_base_url = 'https://www.anapioficeandfire.com/api/books/'
```

Получить информацию обо всех домах Вестероса (Houses) и сохранить в pd.DataFrame

```python
houses_url = 'https://www.anapioficeandfire.com/api/houses/'
filter = {'region':'The Westerlands'}
houses = requests.get(houses_url, filter
).json()
df_houses = pd.DataFrame(houses)
df_houses
```

Получить информацию обо всех домах Вестероса, у которых есть девиз, и сохранить в pd.DataFrame




```python
houses_url = 'https://www.anapioficeandfire.com/api/houses/'
filter = {'region':'The Westerlands'}
houses = requests.get(houses_url, filter
).json()
df_houses = pd.DataFrame(houses)
df_houses = df_houses[df_houses['words']!='']
df_houses
```
