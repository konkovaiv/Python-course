```python
import pandas as pd
import bottleneck as bn
import numpy as np

df_pd = pd.read_csv("train.csv")
```

Посчитайте средний возраст пассажиров и его стандартное отклонение с помощью bottleneck

```python
age = df_pd["Age"].dropna().to_numpy()
mean_age = bn.nanmean(age)
print(mean_age)
```

```python
age = df_pd["Age"].dropna().to_numpy()
mean_age = bn.nanstd(age)
print(mean_age)
```

Для каждого пассажира умножьте значение столбца Fare на 1.3 и сохраните результаты как новый столбец Fare_new

```python
def func(x):
  return x * 1.3
fare_new = df_pd["Fare"].apply(func).reset_index(name="Fare_new"pfprint(fare_new)
```

Оптимизация типов pandas


```python
df = pd.read_csv("Housing.csv")
df.info()
df.describe()
```

```python
df['price'] = df['price'].astype(np.uint32) #целые значения в диапазоне от 1 750 000 до 13 300 000, без отрицательных значений
df['area'] = df['area'].astype(np.uint16) #целые значения от 1650 до 16 200 без отриц.знач
df['bedrooms'] = df['bedrooms'].astype(np.uint8)  #целые значения от 1 до 6 без отриц.знач
df['bathrooms'] = df['bathrooms'].astype(np.uint8)  #целые значения от 1 до 4 без отриц.знач
df['stories'] = df['stories'].astype(np.int8)  #целые значения от 1 до 4 без отриц.знач
df['parking'] = df['parking'].astype(np.int8)  #целые значения от 0 до 3 без отриц.знач
df['mainroad'] = df['mainroad'].astype("category") #присвоен категориальный тип
df['guestroom'] = df['guestroom'].astype("category")  #присвоен категориальный тип
df['basement'] = df['basement'].astype("category")  #присвоен категориальный тип
df['mainroad'] = df['mainroad'].astype("category")  #присвоен категориальный тип
df['hotwaterheating'] = df['hotwaterheating'].astype("category")  #присвоен категориальный тип
df['airconditioning'] = df['airconditioning'].astype("category")  #присвоен категориальный тип
df['prefarea'] = df['prefarea'].astype("category")  #присвоен категориальный тип
df['furnishingstatus'] = df['furnishingstatus'].astype("category")  #присвоен категориальный тип

df.info() #количество используемой памяти снизилось с 55,5КБ до 10,0 КБ
```
