```python
class Chocolate():
  def __init__(self, price: float, addition: str = "Нет"):
    if price <= 0:
      raise ValueError("Некорректная цена.")
    self.price = price
    self.addition = addition

  def show_my_chocolate(self):
    if self.addition == "Нет":
      print("Обычный шоколадный батончик.")
    else:
      print("Шоколадный батончик, добавка: " + self.addition)

  def __str__(self):
    return "Шоколадный батончик, " + str(self.price) + " рублей."

Choco1 = Chocolate(25.0, "лесной орех")
Choco2 = Chocolate(55.22, "с карамелью")
Choco3 = Chocolate(150.20,"Нет")

print(Choco1)
Choco1.show_my_chocolate()
print(Choco2)
Choco2.show_my_chocolate()
print(Choco3)
Choco3.show_my_chocolate()

#Если цена ошибочная
try:
    Error = Chocolate(0.0, "с карамелью")
except ValueError as er:
    print(f"Ошибка: {er}")
```
