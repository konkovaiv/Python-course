**Cамое популярное мужское и самое популярное женское имя на корабле в каждом классе**

```python
#Женское имя
df["Second Name"] = df["Name"].str.split().str[0].str.strip(',')
female_name = (
    df[df['Sex'] == 'female']
    .groupby(['Pclass', 'Second Name'])
    .size()
    .reset_index(name='cnt')
    .sort_values('cnt', ascending=False)
    .groupby('Pclass')
    .first()
    .reset_index()
    [['Pclass', 'Second Name', 'cnt']]
)
print(female_name)
```

```python
#Мужское имя
df["Second Name"] = df["Name"].str.split().str[0].str.strip(',')
male_name = (
    df[df['Sex'] == 'male']
    .groupby(['Pclass', 'Second Name'])
    .size()
    .reset_index(name='cnt')
    .sort_values('cnt', ascending=False)
    .groupby('Pclass')
    .first()
    .reset_index()
    [['Pclass', 'Second Name', 'cnt']]
)
print(male_name)
```

**количества n-местных кабин (в которых было 2, 3, 4, ... человека**
